Introdução:

O que é um banco de dados?

Os bancos de dados são coleçoes organizadas de dados, geralmente armazenados e acessados eletronicamente a partir de um sistema de computador.

Tipos de Bancos de Dados

Existem vários tipos de bancos de dados, incluindo relacionais, não relacionais, orientado à objetos e muito mais. O tipo mais comum é o banco de dados relacional, que organiza os dados em tabelas.

O papel do SGBD

Os sistemas de gerenciamento de banco de dados (SGBD) são softwares que integram com o usuário, outras aplicações e o próprio banco de dados para capturar e analisar os dados. Existem muitos SGBD's disponíveis no mercado, alguns dos mais populares incluem: MySQL, PostgreSQL, SQLite, Oracle Database, Microsoft SQL Server, Redis, MongoDB e MariaDB.

Chaves Primárias

Cada tabela em um banco de dados relacional deve ter uma chave primária. A chave primária é uma coluna(ou conjunto de colunas) cujo valor é único para cada registro. Isso garante que cada registro na tabela possa ser identificado de maneira única. Por exemplo, na tabela "Clientes", o campo "ID" pode ser a chave primária.

Chaves Estrangeiras

Por exemplo, em uma tabela "Pedidos", podemos ter um campo "ClienteID" que seja uma chave estrangeira apontando para a chave primária da tabela "Clientes. Isso cria um relacionamento entre "Pedidos" e "Clientes", permitindo que cada pedido seja associado a um cliente específico.

Relacionamento entre tabelas
exi
Os bandos de daods relacionais permitem estabelecer relações entre tabelas. As relações podem ser "um para um", "um para muitos", ou "muitos para muitos". Estas relações permitem efetuar consultas complexas que unem dados de várias tabelas.

```python
#Conectando com o Banco de Dados SQLite

import sqlite3
from pathlib import Path

ROOT_PATH = Path.cwd() #Note que usei o Path.cwd() para pegar o caminho do diretório atual. Caso seja executado em um script, ideal trocar para Path(__file__).parent 
#para pegar o diretório do script.#

conexao = sqlite3.connect(ROOT_PATH / 'banco_de_dados.sqlite')
print(conexao)
```

```python
#CRIANDO UMA TABELA NO BANCO DE DADOS

#Precisamos criar um cursor para executar comandos SQL. O cursor é um objeto que permite interagir com o banco de dados, executando consultas e manipulando os resultados.

cursor = conexao.cursor()

cursor.execute('CREATE TABLE clientes (id INTEGER PRIMARY KEY AUTOINCREMENT, nome VARCHAR(100), email VARCHAR(150))') # Aqui estamos criando uma tabela chamada "clientes" com três colunas: "id", "nome" e "email". 
#A coluna "id" é definida como chave primária e será incrementada automaticamente a cada novo registro. 
#Note que o nome e email são definidos como VARCHAR, que é um tipo de dado para armazenar strings de tamanho variável.
```

```python
#INSERINDO DADOS NA TABELA

#Agora, vamos criar uma variável chamada "data", que é uma tupla contendo os valores que queremos inserir na tabela.

data = ("Raphael Quintana", "raphael.quintana@gmail.com")

#Agora usamos o método execute() do cursor para executar um comando SQL INSERT, que insere os valores da tupla "data" na tabela "clientes".

cursor.execute('INSERT INTO clientes (nome, email) VALUES (?, ?)', data) #Lembrando que usamos o ? como um marcador de posição para os valores que queremos inserir. 
#Isso ajuda a prevenir ataques de SQL injection.

#Usamos o método commit() da conexão para salvar as alterações no banco de dados.
conexao.commit()
```

```python
#ATUALIZANDO DADOS NA TABELA

#Vamos criar uma variável chamada "data", que é uma tupla contendo os novos valores que queremos atualizar na tabela.

data = ("Raphael Quintana", "raphael.quintana@yahoo.com", 1) #Note que incluímos o id do cliente que queremos atualizar como o último elemento da tupla.    

cursor.execute('UPDATE clientes SET nome = ?, email = ? WHERE id = ?;', data) #Lembrando que usamos o ? como um marcador de posição para os valores que queremos atualizar.
conexao.commit()               
```
